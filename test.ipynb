```python
import geopandas as gpd
import matplotlib.pyplot as plt
from matplotlib.patches import Patch
import mapclassify
from matplotlib_scalebar.scalebar import ScaleBar

name_territory = "Sainte-Baume"
```

```python
grid= gpd.read_file(f"data_output/distance_pop_{name_territory}.shp")
print(grid.columns)
```

```python
fig, ax = plt.subplots(figsize=(10, 8))

grid.plot(ax =ax,column="access_rt", cmap="YlOrRd", scheme="Quantiles",k=6,legend=True)
ax.set_title("Accessibilité à la population", fontsize=18, fontweight="bold")
scalebar = ScaleBar(1, units="m", location="lower right")
ax.add_artist(scalebar)
plt.axis("off")
#ax.set_axis_off()
plt.show()
```

```python
grid.plot(figsize=(12,6),column="access_rt", cmap="YlOrRd", scheme="fisherjenks",k=5,legend=True)
scalebar = ScaleBar(1, units="m", location="lower right")
ax.add_artist(scalebar)
plt.show()
```
